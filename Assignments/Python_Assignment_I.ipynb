# Task 1: Student Marks and Grades


We check the mark ranges from highest to lowest. A student passes with 40 marks or more.
The challenge is to count how many students passed.

```python
students = {
    "Ali": 85,
    "Sara": 72,
    "Ahmed": 35,
    "Ayesha": 91,
    "Usman": 55
}

def calculate_grade(marks):
    if marks >= 80:
        return "A"
    elif marks >= 60:
        return "B"
    elif marks >= 40:
        return "C"
    else:
        return "Fail"

passed_count = 0

for name, marks in students.items():
    grade = calculate_grade(marks)
    print(f"{name}: Marks = {marks}, Grade = {grade}")

    if marks >= 40:
        passed_count += 1

print("Number of students passed:", passed_count)
```

**How it works:** `calculate_grade()` returns the grade for a mark. The `for` loop visits every student. The counter increases only when marks are at least 40.

**Expected result:** Ali A, Sara B, Ahmed Fail, Ayesha A, Usman C; 4 students passed.

# Task 2: Shopping Bill Calculator

**Concepts:** dictionary, `for` loop, function, `if-else`, `while` loop.

The customer can type product names repeatedly and type `done` to finish. Product matching is case-insensitive. Unknown products are reported and not added to the bill. A 10% discount applies when the original bill is Rs. 5,000 or more.

```python
products = {
    "Milk": 250,
    "Bread": 150,
    "Eggs": 400,
    "Rice": 1200
}

def calculate_discount(total):
    if total >= 5000:
        return total * 0.10
    else:
        return 0

print("Available products:")
for item, price in products.items():
    print(f"{item}: Rs. {price}")

total = 0

while True:
    item = input("Enter a product name, or 'done' to finish: ").strip().title()

    if item == "Done":
        break

    if item in products:
        # Ask for quantity so a customer can buy more than one unit.
        try:
            quantity = int(input(f"Quantity of {item}: "))
            if quantity <= 0:
                print("Quantity must be greater than zero.")
                continue
            total += products[item] * quantity
        except ValueError:
            print("Please enter quantity as a whole number.")
    else:
        print("Product not found. Please choose an item from the list.")

discount = calculate_discount(total)
final_bill = total - discount

print(f"Original bill: Rs. {total:.2f}")
print(f"Discount: Rs. {discount:.2f}")
print(f"Final bill: Rs. {final_bill:.2f}")
```

**How it works:** The loop keeps asking for products until `done`. The function returns 10% of the total if it reaches Rs. 5,000; otherwise it returns zero. Quantity is included as a useful extension to the worksheet.

# Task 3: ATM Withdrawal System

**Concepts:** `while` loop, nested `if-else`, functions, dictionary.

The account starts with a balance of Rs. 10,000 and PIN 1234. The user gets at most three PIN attempts. After a correct PIN, the program asks for a withdrawal amount. It rejects non-positive amounts and withdrawals larger than the balance.

```python
account = {
    "name": "Ali",
    "pin": 1234,
    "balance": 10000
}

def withdraw(balance, amount):
    if amount <= 0:
        print("Withdrawal amount must be greater than zero.")
        return balance
    elif amount <= balance:
        balance -= amount
        print("Withdrawal successful.")
        return balance
    else:
        print("Insufficient balance.")
        return balance

attempts = 0
logged_in = False

while attempts < 3:
    pin_text = input("Enter your PIN: ").strip()

    try:
        pin = int(pin_text)
    except ValueError:
        pin = None

    if pin == account["pin"]:
        logged_in = True
        print(f"Welcome, {account['name']}!")
        break
    else:
        attempts += 1
        remaining = 3 - attempts
        if remaining > 0:
            print(f"Incorrect PIN. Attempts remaining: {remaining}")
        else:
            print("Your account session is locked after three incorrect attempts.")

if logged_in:
    try:
        amount = float(input("Enter withdrawal amount (Rs.): "))
        account["balance"] = withdraw(account["balance"], amount)
        print(f"Remaining balance: Rs. {account['balance']:.2f}")
    except ValueError:
        print("Please enter a valid numeric amount.")
```

**How it works:** The outer loop limits PIN attempts. A correct PIN ends the loop with `break`. Only after successful login does the nested withdrawal logic run. The `withdraw()` function returns the updated balance or leaves it unchanged when the withdrawal is invalid.

# Task 4: Number Guessing Game

**Concepts:** `while` loop, `if-elif-else`, functions.

The secret number is 7. The player gets five attempts. The game stops when the guess is correct or the attempts run out. The challenge stores the player's name and score in a dictionary; here, score means the number of correct guesses (1 for a win, 0 otherwise).

```python
def check_guess(guess, secret):
    if guess == secret:
        return "Correct!"
    elif guess < secret:
        return "Too low!"
    else:
        return "Too high!"

secret_number = 7
attempts = 0
won = False
player_name = input("Enter your name: ").strip() or "Player"

while attempts < 5:
    try:
        guess = int(input("Guess the number (1-10): "))
    except ValueError:
        print("Please enter a whole number.")
        continue

    attempts += 1
    result = check_guess(guess, secret_number)
    print(result)

    if guess == secret_number:
        won = True
        break

score = 1 if won else 0
player = {"name": player_name, "score": score, "attempts_used": attempts}

if won:
    print(f"Well done, {player_name}! You guessed it in {attempts} attempt(s).")
else:
    print("No attempts left. The secret number was", secret_number)

print("Player record:", player)
```

**How it works:** The function compares the guess with 7 and returns a message. Each valid integer guess uses one attempt. A correct guess sets `won` to `True` and ends the loop.

# Task 5: Restaurant Ordering System

**Concepts:** dictionary, `while` loop, nested `if-else`, functions.

The customer can order several menu items until typing `done`. If an item exists, the program asks for its quantity and adds its price multiplied by quantity. The challenge is solved by keeping ordered items and quantities in a separate dictionary.

```python
menu = {
    "Biryani": 350,
    "Karahi": 1200,
    "Naan": 50,
    "Tea": 100
}

def calculate_item_total(price, quantity):
    return price * quantity

ordered_items = {}
total = 0

print("Restaurant Menu:")
for item, price in menu.items():
    print(f"{item}: Rs. {price}")

while True:
    item = input("Enter item name or 'done': ").strip().title()

    if item == "Done":
        break

    if item in menu:
        try:
            quantity = int(input(f"Enter quantity of {item}: "))
            if quantity <= 0:
                print("Quantity must be greater than zero.")
            else:
                item_total = calculate_item_total(menu[item], quantity)
                total += item_total
                ordered_items[item] = ordered_items.get(item, 0) + quantity
                print(f"{quantity} x {item} = Rs. {item_total}")
        except ValueError:
            print("Please enter quantity as a whole number.")
    else:
        print("Sorry, that item is not on the menu.")

print("\nOrder summary:")
if ordered_items:
    for item, quantity in ordered_items.items():
        print(f"{item}: {quantity} x Rs. {menu[item]} = Rs. {menu[item] * quantity}")
else:
    print("No items ordered.")

print(f"Total bill: Rs. {total}")
print("Ordered items and quantities:", ordered_items)
```

**How it works:** `ordered_items.get(item, 0)` returns the existing quantity, or zero if the item has not been ordered yet. This lets repeated orders of the same item combine into one entry.

# Task 6: Employee Attendance and Salary

**Concepts:** dictionary, `for` loop, `if-elif-else`, functions.

Monthly salary is Rs. 30,000. Exactly 26 days earns 100%; 20–25 days earns 80%; below 20 days earns 50%; more than 26 days is invalid. The challenge counts employees who received full salary.

```python
employees = {
    "Ali": 26,
    "Sara": 23,
    "Ahmed": 18,
    "Ayesha": 30
}

MONTHLY_SALARY = 30000

def calculate_salary(days):
    if days > 26:
        return None  # None marks invalid attendance
    elif days == 26:
        return MONTHLY_SALARY
    elif days >= 20:
        return MONTHLY_SALARY * 0.80
    elif days >= 0:
        return MONTHLY_SALARY * 0.50
    else:
        return None  # Negative attendance is invalid

full_salary_count = 0

for name, days in employees.items():
    salary = calculate_salary(days)

    if salary is None:
        print(f"{name}: Invalid attendance ({days} days)")
    else:
        print(f"{name}: Attendance = {days} days, Salary = Rs. {salary:.2f}")
        if days == 26:
            full_salary_count += 1

print("Employees who received full salary:", full_salary_count)
```

**How it works:** The function returns `None` for invalid attendance, so the main loop can print an error instead of a salary. The full-salary counter increases only when attendance is exactly 26 days.
